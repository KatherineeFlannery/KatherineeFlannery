```python
#Install faker
!pip install faker

#Import libraries
import pandas as pd
import numpy as np
import random
from random import randint
from faker import Faker

#Set faker seed to 4321
fake = Faker()
Faker.seed(4321)

#Set data ranges
scoring_list = [50, 58, 66, 74, 82, 88, 94, 100]
gender = ['Male', 'Female']
race = ['BlackOrAfricanAmerican', 'White', 'Asian', 'NativeHawaiian', 'AmericanIndian']
Hispanic = ['Yes', 'No']

#Create dataframe with synthetic data
def input_data(x):
    data = []
    for _ in range(x):
        data.append({
            'StudentID': randint(10000, 99999),
            'StudentName': fake.name(),
	    'ClassPeriod': random.randint(1, 6),
            'Gender': random.choice(gender),
            'RaceEthnicity': random.choice(race),
            'Hispanic': random.choice(Hispanic),
            'SBAC': random.randint(1, 4),
	    'iReadyScore': random.randint(350, 580),
            '7GA1': random.choice(scoring_list),
            '7RP2AD': random.choice(scoring_list),
            '7RPA2A': random.choice(scoring_list),
            '7RPA2B': random.choice(scoring_list),
            '7RPA2C': random.choice(scoring_list),
            '7GB4': random.choice(scoring_list)
        })
    return pd.DataFrame(data)

df = input_data(110)

print(df.head())

#Export to CSV file
df.to_csv("Generated_Student_Data.csv")
```
